# EDA: NASA C-MAPSS FD001

Este notebook vuelve a ejecutar el análisis exploratorio desde los archivos originales. Las etiquetas RUL oficiales de test se usan únicamente para describir el último ciclo observado de cada motor. Ningún archivo crudo se modifica.

```python
from pathlib import Path
from IPython.display import Image, display
import yaml

from predictguard.data.eda import run_eda
from predictguard.data.validate import resolve_data_dir

repo_root = Path.cwd()
while not (repo_root / "configs" / "config.yaml").is_file() and repo_root != repo_root.parent:
    repo_root = repo_root.parent
config_path = repo_root / "configs" / "config.yaml"
config = yaml.safe_load(config_path.read_text(encoding="utf-8"))
outputs = run_eda(
    resolve_data_dir(config_path),
    repo_root / "artifacts" / "figures",
    repo_root / "artifacts" / "metrics",
    repo_root / "docs" / "eda_findings.md",
    window_size=int(config["window"]["size"]),
)
print(f"Reporte: {outputs['report_path']}")
```

```python
display(outputs["sensor_summary"].sort_values("variance"))
display(outputs["strong_correlations"])
display(outputs["rul_balance"])
print(outputs["sequence_diagnostics"])
```

```python
figure_names = [
    "eda_engine_durations.png",
    "eda_sensor_trajectories.png",
    "eda_sensor_variance.png",
    "eda_sensor_correlations.png",
    "eda_train_rul_distribution.png",
    "eda_train_test_sensor_distributions.png",
    "eda_lifecycle_segments.png",
]
for figure_name in figure_names:
    figure_path = outputs["figure_dir"] / figure_name
    display(Image(filename=str(figure_path)))
```
